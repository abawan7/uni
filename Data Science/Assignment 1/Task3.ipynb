```python
import time
import pandas as pd
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from bs4 import BeautifulSoup
from selenium.webdriver.common.by import By
```

```python
url = 'https://www.zameen.com/Houses_Property/Lahore_DHA_Defence-9-1.html'

ptions = webdriver.ChromeOptions()
options = webdriver.ChromeOptions()
options.add_argument('--headless')  # Run in headless mode
options.add_argument('--no-sandbox')
options.add_argument('--disable-dev-shm-usage')
options.binary_location = '/usr/bin/google-chrome'  # Set the Chrome binary location
service = Service(ChromeDriverManager().install())

# Initialize the WebDriver with Chrome
driver = webdriver.Chrome(service=service, options=options)

url = 'https://www.zameen.com/Houses_Property/Lahore_DHA_Defence-9-1.html'
driver.get(url)
```

```python
all_properties = []
for page_num in range(1, 6):
  if page_num > 1:
        url = f'https://www.zameen.com/Houses_Property/Lahore_DHA_Defence-9-{page_num}.html'
        driver.get(url)
        time.sleep(3)

  soup = BeautifulSoup(driver.page_source, 'html.parser')
  property_list = soup.find(class_='e20beb46')

  listings = property_list.find_all(class_='a37d52f0')

  for listing in listings:

    price = listing.find(class_='dc381b54').text.strip()
    location = listing.find(class_='db1aca2f').text.strip()
    beds_element = listing.find('span', class_='_6d9b9b83', attrs={'aria-label': 'Beds'})
    if beds_element:
      beds = beds_element.text.strip()
    else:
      beds = 'N/A'

    baths_element = listing.find('span', class_='_6d9b9b83', attrs={'aria-label': 'Baths'})
    if baths_element:
       baths = baths_element.text.strip()
    else:
      baths = 'N/A'
    area_divs = listing.find_all(class_='cb0c0514')
    area = area_divs[1].find('span').text.strip()

    all_properties.append([location, price, beds, baths, area])


driver.quit()
```

```python
df = pd.DataFrame({
    'Location': [prop[0] for prop in all_properties],
    'Price': [prop[1] for prop in all_properties],
    'Beds': [prop[2] for prop in all_properties],
    'Baths': [prop[3] for prop in all_properties],
    'Area': [prop[4] for prop in all_properties]
})

print(df)
```
