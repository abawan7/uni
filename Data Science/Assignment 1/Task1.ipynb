```python
import requests
from bs4 import BeautifulSoup
```

```python
url = 'https://www.codetechmentor.com/p/about-me.html'
html = requests.get(url)
```

```python
profile = BeautifulSoup(html.content, 'html.parser')
details = profile.find(class_='intro')
name = details.find('h1').find(class_='highlight')
introductions = details.find_all('p')
print(name.text.strip())
for intro in introductions:
  print(intro.text.strip())

social_media = profile.find(class_='social-icons')
links = social_media.find_all('a')
for link in links:
  print(link.get('href'))
```
