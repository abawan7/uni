```python
import requests
from bs4 import BeautifulSoup
```

```python
url = 'https://priceoye.pk/laptops/apple'
html = requests.get(url)
```

```python
content = BeautifulSoup(html.content, 'html.parser')
product_list = content.find(class_='product-list')
product_boxes = product_list.find_all(class_='productBox', attrs={'data-brand': 'apple'})

for product_box in product_boxes:
    check_discontinue = product_box.find(class_="ribbon ribbon_discontinued")
    if check_discontinue:
      continue
    else:
      print(product_box.find(class_='p-title').text.strip())
      print(product_box.find(class_='price-box').text.strip())
      discount = product_box.find(class_='price-diff-saving')
      if discount:
        print(discount.text.strip())
      else:
        print('No discount found')

    print("\n")
```
